```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df=pd.read_csv('Titanic-Dataset.csv')[['Age', 'Pclass', 'SibSp', 'Parch','Survived']]
```

```python
df.head()
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

```python
X=df.iloc[:,0:-1]
y=df.iloc[:,-1]
```

```python
X.head()
```

```python
y.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=20))
```

# Applying the Family Construction

```python
df.columns
```

```python
X['Family_Size']=X['Parch']+X['SibSp']+1
```

```python
X.head()
```

```python
def myfunc(num):
    if num==1:
        return 0
    elif num>1 and num<=4:
        return 1
    else:
        return 2
```

```python
X['Family_type']=X['Family_Size'].apply(myfunc)
```

```python
X.head()
```

```python
X.drop(columns=['Parch', 'SibSp', 'Family_Size'], inplace=True)
```

```python
X.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), X, y, scoring='accuracy', cv=20))
```

# **Feature Splitting **

```python
df=pd.read_csv('Titanic-Dataset.csv')
```

```python
df['Name']
```

```python
df['Title']=df['Name'].str.split(',',expand=True)[1].str.split('.', expand=True)[0]
```
