```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset(name='iris')
df.head()
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['species']=le.fit_transform(df['species'])
```

```python
df=df[["sepal_length", "petal_length", "species"]]
```

```python
df.head()
```

```python
X=df.iloc[:,0:2]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3 , random_state=13)
```

```python
from sklearn.linear_model import LogisticRegression
lr= LogisticRegression()
lr.fit(X_train, y_train)
```

```python
y_pred= lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("accuracy_score: ",accuracy_score(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix
pd.DataFrame(confusion_matrix(y_test , y_pred))
```

```python
query=np.array([[5.1, 1.4]])
lr.predict_proba(query)
```

```python
lr.predict(query)
```

```python
plt.scatter(X.iloc[:,0], X.iloc[:,1], c=y)
plt.show()
```

```python
x_min, x_max = X.iloc[:,0].min() - 1, X.iloc[:,0].max() + 1
y_min, y_max = X.iloc[:,1].min() - 1, X.iloc[:,1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))
```

```python
Z = lr.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)
```

```python
plt.contourf(xx, yy, Z, alpha=0.3)   # decision regions
plt.scatter(X.iloc[:,0], X.iloc[:,1], c=y, edgecolor="k")  # data points
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Multinomial Logistic Regression (Softmax)")
plt.show()
```
