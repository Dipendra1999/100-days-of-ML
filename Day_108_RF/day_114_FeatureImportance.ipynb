```python
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier

```

```python
X,y=make_classification(n_samples=5, n_classes=2, n_features=2, n_informative=2, n_redundant=0, random_state=0)
```

```python
dt=DecisionTreeClassifier()
dt.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(dt)
```

```python
dt.feature_importances_
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=2)
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
rf.estimators_
```

```python
print(rf.estimators_[0].feature_importances_)
print(rf.estimators_[1].feature_importances_)

print((0.555555 + 0.7916667)/2)
```

### Same Example with more number of samples

```python
X,y=make_classification(n_samples=15, n_classes=2, n_features=2, n_informative=2, n_redundant=0,random_state=29)
clf=DecisionTreeClassifier()
clf.fit(X,y)
print(plot_tree(clf))
print(clf.feature_importances_)
```
