```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
X=np.random.rand(100,1) - 0.5
y=3*X[:,0]**2 + 0.05*np.random.randn(100)
```

```python
X
```

```python
y
```

```python
df=pd.DataFrame({'X':X.ravel(),"y":y})
```

```python
df
```

```python
plt.scatter(df['X'],df['y'])
plt.title("X vs y")
```

now Here i am creating the 3 model 
1st model is  M1=(mean)
2nd Model is Decision tree
3rd Model is Decision Tree

```python
df['pred1']=df['y'].mean()
```

```python
df
```

```python
# here res1 means the pseudo-residual of the model 1 predictions from the actual 
df['res1']=df['y']-df['pred1']
```

```python
df
```

```python
plt.scatter(df['X'],df['y'])
plt.plot(df['X'],df['pred1'],color='red')
```

```python
from sklearn.tree import DecisionTreeRegressor
tree1=DecisionTreeRegressor(max_leaf_nodes=8)
tree1.fit(df['X'].values.reshape(100,1),df['res1'].values)
```

```python
from sklearn.tree import plot_tree
plot_tree(tree1)
plt.show()
```

```python
#generating the X test data
X_test=np.linspace(-0.5,0.5, 500)
y_pred= 0.265458 + tree1.predict(X_test.reshape(500,1))
```

```python
plt.figure(figsize=(10,5))
plt.plot(X_test,y_pred,linewidth=2, color='red')
plt.scatter(df['X'],df['y'])
plt.show()
```

```python
df['pred2']= 0.265458 + tree1.predict(df['X'].values.reshape(100,1))
```

```python
df
```

```python
df['res2']=df['y']-df['pred2']
```

```python
df
```

```python
tree2=DecisionTreeRegressor(max_leaf_nodes=8)
tree2.fit(df['X'].values.reshape(100,1),df['res2'].values)
```

```python
y_pred=0.265458 + sum(regressor.predict(X_test.reshape(-1,1)) for regressor in [tree1, tree2])
```

```python
plt.figure(figsize=(10,4))
plt.scatter(df['X'],df['y'])
plt.plot(X_test,y_pred,linewidth=2,color='red')
plt.title('X vs y')
```
