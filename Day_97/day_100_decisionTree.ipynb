```python
import graphviz.backend as be
from sklearn.tree import DecisionTreeClassifier
from dtreeviz import model
from IPython.display import display_svg, Image, SVG
from sklearn.tree import plot_tree
from sklearn.tree import export_graphviz
```

```python
import pandas as pd
import numpy as np
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()

X_train= iris.data
y_train=iris.target
```

```python
dt=DecisionTreeClassifier(max_depth=2)
```

```python
dt.fit(X_train, y_train)
```

```python
plot_tree(dt,feature_names=iris.feature_names,class_names=iris.target_names,filled=True)
```

```python
viz_model_iris=model(model = dt,
             X_train=X_train,
             y_train=y_train,
             feature_names=iris.feature_names,
             class_names=iris.target_names,
             target_name="species",
             )

```

```python
viz_iris = viz_model_iris.view(scale=1.5)
```

```python
viz_iris
```

```python
x=iris.data[np.random.randint(0,len(iris.data)),:]
viz_iris2=viz_model_iris.view(x=x)
viz_iris2
```

## Regression Tree

```python
from sklearn.datasets import load_wine
```

```python
wine = load_wine()
```

```python
X=wine.data
y=wine.target
```

```python
X
```

```python
df=pd.DataFrame(X,columns=wine.feature_names)
```

```python
wine.target_names
```

```python
df['target']=wine.target
```

```python
df.info()
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
regr=DecisionTreeRegressor(max_depth=3)
X_train=wine.data
y_train=wine.target
regr.fit(X_train, y_train)

viz_model = model(regr,
                  X_train,
                  y_train,
                  target_name='target',
                  feature_names=wine.feature_names,
                  class_names=wine.target_names,
                  )
```

```python
viz=viz_model.view(orientation="LR", show_node_labels=True, scale=1.5)
```

```python
viz
```
